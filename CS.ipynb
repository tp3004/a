### Q1 — Question 1

**Write a program to implement Breadth First Search (BFS) for the Monkey Banana Problem.**

```python
from collections import deque

def monkey_banana_bfs():

    start = ("door", "middle", False, False)
    queue = deque([(start, [])])
    visited = {start}

    while queue:
        state, path = queue.popleft()
        monkey_pos, box_pos, on_box, banana = state

        if banana:
            print("Solution found:")
            for step in path:
                print(step)
            return

        next_states = []

        for new_pos in ["door", "middle", "window"]:
            if new_pos != monkey_pos:
                next_states.append(
                    ((new_pos, box_pos, False, banana),
                     f"Monkey moves from {monkey_pos} to {new_pos}.")
                )

        if monkey_pos == box_pos:
            for new_box_pos in ["door", "window"]:
                if new_box_pos != box_pos:
                    next_states.append(
                        ((new_box_pos, new_box_pos, False, banana),
                         f"Monkey pushes the box from {box_pos} to {new_box_pos}.")
                    )

        if monkey_pos == box_pos and not on_box:
            next_states.append(
                ((monkey_pos, box_pos, True, banana),
                 "Monkey climbs onto the box.")
            )

        if on_box and monkey_pos == "middle":
            next_states.append(
                ((monkey_pos, box_pos, True, True),
                 "Monkey grabs the banana.")
            )

        for new_state, action in next_states:
            if new_state not in visited:
                visited.add(new_state)
                queue.append((new_state, path + [action]))

    print("No solution found.")

monkey_banana_bfs()
```

### Q2 — Question 2

**Write a program to implement the A* Search Algorithm to find the shortest path between a source node and a destination node.**

```python
import heapq

def a_star(graph, heuristic, start, goal):
    priority_queue = [(heuristic[start], 0, start, [start])]
    best_cost = {start: 0}

    while priority_queue:
        f, g, node, path = heapq.heappop(priority_queue)

        if node == goal:
            return path, g

        for neighbor, cost in graph.get(node, []):
            new_g = g + cost

            if neighbor not in best_cost or new_g < best_cost[neighbor]:
                best_cost[neighbor] = new_g
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(
                    priority_queue,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None, float("inf")

graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("A", 1), ("C", 2), ("D", 5)],
    "C": [("A", 4), ("B", 2), ("D", 1), ("E", 3)],
    "D": [("B", 5), ("C", 1), ("E", 2), ("F", 5)],
    "E": [("C", 3), ("D", 2), ("F", 2)],
    "F": [("D", 5), ("E", 2)]
}

heuristic = {
    "A": 7, "B": 6, "C": 4,
    "D": 3, "E": 2, "F": 0
}

start = input("Enter source node: ").upper()
goal = input("Enter destination node: ").upper()

path, cost = a_star(graph, heuristic, start, goal)

if path:
    print("Shortest path:", " -> ".join(path))
    print("Total cost:", cost)
else:
    print("No path exists.")
```

### Q2 (OR) — Question 3

**Write a program to solve the Map Coloring Problem using the Constraint Satisfaction Problem (CSP) approach.**

```python
def is_safe(node, color, assignment, graph):
    for neighbor in graph[node]:
        if assignment.get(neighbor) == color:
            return False
    return True

def solve_csp(nodes, colors, graph, assignment):
    if len(assignment) == len(nodes):
        return True

    node = next(n for n in nodes if n not in assignment)

    for color in colors:
        if is_safe(node, color, assignment, graph):
            assignment[node] = color

            if solve_csp(nodes, colors, graph, assignment):
                return True

            del assignment[node]

    return False

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

colors = ["Red", "Green", "Blue"]
assignment = {}

if solve_csp(list(graph.keys()), colors, graph, assignment):
    print("Map Coloring Solution:")
    for region, color in assignment.items():
        print(region, "->", color)
else:
    print("No solution exists.")
```

**Write a program to solve the Water Jug Problem using BFS.**

```python
from collections import deque

def water_jug_bfs(capacity_a, capacity_b, target):
    start = (0, 0)
    queue = deque([(start, [])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()

        if a == target or b == target:
            print("Solution steps:")
            for state in path + [(a, b)]:
                print(state)
            return

        states = [
            ((capacity_a, b), "Fill Jug A"),
            ((a, capacity_b), "Fill Jug B"),
            ((0, b), "Empty Jug A"),
            ((a, 0), "Empty Jug B"),
            ((a - min(a, capacity_b - b), b + min(a, capacity_b - b)),
             "Pour Jug A into Jug B"),
            ((a + min(b, capacity_a - a), b - min(b, capacity_a - a)),
             "Pour Jug B into Jug A")
        ]

        for new_state, action in states:
            if new_state not in visited:
                visited.add(new_state)
                queue.append((new_state, path + [(new_state, action)]))

    print("No solution exists.")

capacity_a = int(input("Enter capacity of Jug A: "))
capacity_b = int(input("Enter capacity of Jug B: "))
target = int(input("Enter target amount: "))

water_jug_bfs(capacity_a, capacity_b, target)
```

### Q2 — Question 5

**Write a program to implement the Backtracking Algorithm for solving a constraint satisfaction problem.**

```python
def is_safe(var, value, assignment, constraints):
    for other in constraints[var]:
        if assignment.get(other) == value:
            return False
    return True

def backtrack(variables, domains, constraints, assignment):
    if len(assignment) == len(variables):
        return assignment.copy()

    unassigned = [v for v in variables if v not in assignment]
    var = unassigned[0]

    for value in domains[var]:
        if is_safe(var, value, assignment, constraints):
            assignment[var] = value
            result = backtrack(variables, domains, constraints, assignment)

            if result:
                return result

            del assignment[var]

    return None

variables = ["A", "B", "C", "D"]
domains = {v: [1, 2, 3] for v in variables}

constraints = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

solution = backtrack(variables, domains, constraints, {})

if solution:
    print("CSP Solution:")
    for variable, value in solution.items():
        print(variable, "=", value)
else:
    print("No solution exists.")
```

**Write a program to implement BFS traversal for a graph.**

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)
    order = []

    while queue:
        vertex = queue.popleft()
        order.append(vertex)

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "F"],
    "F": ["C", "E"]
}

start = input("Enter starting vertex: ").upper()
print("BFS Traversal:", bfs(graph, start))
```

### Q2 — Question 8

**Write a program to implement Alpha-Beta Pruning for optimizing the Minimax search process.**

```python
def minimax_alpha_beta(depth, node_index, maximizing, values, alpha, beta):
    if depth == 0:
        return values[node_index]

    if maximizing:
        best = float("-inf")

        for i in range(2):
            value = minimax_alpha_beta(
                depth - 1, node_index * 2 + i,
                False, values, alpha, beta
            )
            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    best = float("inf")

    for i in range(2):
        value = minimax_alpha_beta(
            depth - 1, node_index * 2 + i,
            True, values, alpha, beta
        )
        best = min(best, value)
        beta = min(beta, best)

        if beta <= alpha:
            break

    return best

values = [3, 5, 6, 9, 1, 2, 0, -1]
depth = 3

result = minimax_alpha_beta(
    depth, 0, True, values,
    float("-inf"), float("inf")
)

print("Optimal value using Alpha-Beta Pruning:", result)
```

### Q2 (OR) — Question 15

**Write a program to develop a Tic-Tac-Toe Game in which the computer uses the Minimax Algorithm to play optimally.**

```python
def print_board(board):
    for i in range(0, 9, 3):
        print(board[i], "|", board[i+1], "|", board[i+2])
        if i < 6:
            print("--+---+--")

def winner(board, player):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    return any(all(board[i] == player for i in line) for line in wins)

def minimax(board, maximizing):
    if winner(board, "O"):
        return 1
    if winner(board, "X"):
        return -1
    if " " not in board:
        return 0

    if maximizing:
        best = -float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                best = max(best, minimax(board, False))
                board[i] = " "
        return best

    best = float("inf")
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            best = min(best, minimax(board, True))
            board[i] = " "
    return best

def best_move(board):
    best_score = -float("inf")
    move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(board, False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move

board = [" "] * 9

for turn in range(9):
    print_board(board)

    if turn % 2 == 0:
        position = int(input("Enter your position (1-9): ")) - 1
        if position < 0 or position > 8 or board[position] != " ":
            print("Invalid position.")
            continue
        board[position] = "X"

        if winner(board, "X"):
            print_board(board)
            print("You win!")
            break
    else:
        move = best_move(board)
        board[move] = "O"
        print("Computer selected:", move + 1)

        if winner(board, "O"):
            print_board(board)
            print("Computer wins!")
            break
else:
    print_board(board)
    print("Game Draw.")
```

### Q1 — Question 16

**Write a program to solve the 4-Puzzle Problem using the A* Search Algorithm.**

```python
import heapq

GOAL = (1, 2, 3, 4)

def heuristic(state):
    return sum(state[i] != GOAL[i] for i in range(4))

def neighbors(state):
    result = []
    for i in range(4):
        for j in range(i + 1, 4):
            new_state = list(state)
            new_state[i], new_state[j] = new_state[j], new_state[i]
            result.append(tuple(new_state))
    return result

def solve(start):
    pq = [(heuristic(start), 0, start, [start])]
    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state == GOAL:
            return path

        if state in visited:
            continue

        visited.add(state)

        for nxt in neighbors(state):
            if nxt not in visited:
                new_g = g + 1
                heapq.heappush(
                    pq,
                    (new_g + heuristic(nxt), new_g, nxt, path + [nxt])
                )

    return None

start = tuple(map(int, input("Enter 4 puzzle values separated by spaces: ").split()))

if sorted(start) != [1, 2, 3, 4]:
    print("Enter each of 1, 2, 3, 4 exactly once.")
else:
    path = solve(start)
    if path:
        print("A* solution:")
        for state in path:
            print(state)
    else:
        print("No solution.")
```

### Q2 — Question 17

**Write a program or report that compares Minimax and Alpha-Beta Pruning on the same game tree.**

```python
def minimax(values, depth, index, maximizing):
    if depth == 0:
        return values[index]

    if maximizing:
        return max(
            minimax(values, depth - 1, index * 2, False),
            minimax(values, depth - 1, index * 2 + 1, False)
        )

    return min(
        minimax(values, depth - 1, index * 2, True),
        minimax(values, depth - 1, index * 2 + 1, True)
    )

def alpha_beta(values, depth, index, maximizing, alpha, beta, counter):
    counter[0] += 1

    if depth == 0:
        return values[index]

    if maximizing:
        best = float("-inf")
        for child in [index * 2, index * 2 + 1]:
            best = max(
                best,
                alpha_beta(values, depth - 1, child, False, alpha, beta, counter)
            )
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best

    best = float("inf")
    for child in [index * 2, index * 2 + 1]:
        best = min(
            best,
            alpha_beta(values, depth - 1, child, True, alpha, beta, counter)
        )
        beta = min(beta, best)
        if beta <= alpha:
            break
    return best

values = [3, 5, 6, 9, 1, 2, 0, -1]
depth = 3

minimax_value = minimax(values, depth, 0, True)
counter = [0]
alpha_beta_value = alpha_beta(
    values, depth, 0, True,
    float("-inf"), float("inf"), counter
)

print("Minimax value:", minimax_value)
print("Alpha-Beta value:", alpha_beta_value)
print("Alpha-Beta nodes evaluated:", counter[0])
print("Both algorithms return the same optimal value.")
```

### Q1 — Question 19

**Write a program to demonstrate the Hill Climbing Algorithm for finding the maximum value of a simple objective function.**

```python
def f(x):
    return -x**2 + 4*x

def hill_climbing(start, step=0.1):
    current = start

    while True:
        current_value = f(current)
        left = current - step
        right = current + step

        best = current

        if f(left) > f(best):
            best = left

        if f(right) > f(best):
            best = right

        if f(best) <= current_value:
            return current, current_value

        current = best

start = float(input("Enter starting value of x: "))
x, value = hill_climbing(start)

print("Local maximum x =", round(x, 4))
print("Maximum value =", round(value, 4))
```

### Q2 — Question 20

**Write a program to implement the Forward Chaining inference mechanism.**

```python
facts = {"rain", "clouds"}
rules = [
    ({"rain"}, "wet_ground"),
    ({"wet_ground", "clouds"}, "umbrella_needed"),
    ({"umbrella_needed"}, "carry_umbrella")
]

def forward_chaining(facts, rules):
    facts = set(facts)
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if conditions.issubset(facts) and conclusion not in facts:
                facts.add(conclusion)
                changed = True
                print("Derived:", conclusion)

    return facts

print("Initial facts:", facts)
final_facts = forward_chaining(facts, rules)
print("Final knowledge:", final_facts)
```

**Write a program or report that compares Minimax and Alpha-Beta Pruning on the same game tree.**

```python
def minimax(values, depth, index, maximizing):
    if depth == 0:
        return values[index]

    if maximizing:
        return max(
            minimax(values, depth - 1, index * 2, False),
            minimax(values, depth - 1, index * 2 + 1, False)
        )

    return min(
        minimax(values, depth - 1, index * 2, True),
        minimax(values, depth - 1, index * 2 + 1, True)
    )

def alpha_beta(values, depth, index, maximizing, alpha, beta, counter):
    counter[0] += 1

    if depth == 0:
        return values[index]

    if maximizing:
        best = float("-inf")
        for child in [index * 2, index * 2 + 1]:
            best = max(
                best,
                alpha_beta(values, depth - 1, child, False, alpha, beta, counter)
            )
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best

    best = float("inf")
    for child in [index * 2, index * 2 + 1]:
        best = min(
            best,
            alpha_beta(values, depth - 1, child, True, alpha, beta, counter)
        )
        beta = min(beta, best)
        if beta <= alpha:
            break
    return best

values = [3, 5, 6, 9, 1, 2, 0, -1]
depth = 3

minimax_value = minimax(values, depth, 0, True)
counter = [0]
alpha_beta_value = alpha_beta(
    values, depth, 0, True,
    float("-inf"), float("inf"), counter
)

print("Minimax value:", minimax_value)
print("Alpha-Beta value:", alpha_beta_value)
print("Alpha-Beta nodes evaluated:", counter[0])
print("Both algorithms return the same optimal value.")
```

**Write a program to solve a state-space search problem using DFS. Start vertex = 1.**

```python
from collections import deque

def dfs(graph, start):
    visited = set()
    stack = [start]
    order = []

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.add(node)
            order.append(node)

            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return order

graph = {
    1: [2, 3],
    2: [4, 5],
    3: [6],
    4: [],
    5: [],
    6: []
}

print("DFS Traversal from vertex 1:")
print(dfs(graph, 1))
```

```python
def depth_limited_search(graph, start, goal, limit):
    def dls(node, depth, path):
        if node == goal:
            return path

        if depth == 0:
            return None

        for neighbor in graph[node]:
            if neighbor not in path:
                result = dls(neighbor, depth - 1, path + [neighbor])
                if result:
                    return result

        return None

    return dls(start, limit, [start])

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}

start = input("Enter start node: ").upper()
goal = input("Enter goal node: ").upper()
limit = int(input("Enter depth limit: "))

path = depth_limited_search(graph, start, goal, limit)

if path:
    print("Path found:", " -> ".join(path))
else:
    print("Goal not found within the depth limit.")
```

**Write a program to implement the Best First Search Algorithm using heuristic values to find a path from a start node to a goal node.**

```python
import heapq

def best_first_search(graph, heuristic, start, goal):
    priority_queue = [(heuristic[start], start, [start])]
    visited = set()

    while priority_queue:
        h, node, path = heapq.heappop(priority_queue)

        if node == goal:
            return path

        if node in visited:
            continue

        visited.add(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbor], neighbor, path + [neighbor])
                )

    return None

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": ["G"],
    "E": ["G"],
    "F": ["G"],
    "G": []
}

heuristic = {
    "A": 6, "B": 4, "C": 5,
    "D": 2, "E": 3, "F": 1, "G": 0
}

start = input("Enter start node: ").upper()
goal = input("Enter goal node: ").upper()

path = best_first_search(graph, heuristic, start, goal)

if path:
    print("Path found:", " -> ".join(path))
else:
    print("No path found.")
```

### Q2 — Question 35

**Write a program to solve the N-Queens Problem using the Backtracking Algorithm.**

```python
def is_safe(board, row, col, n):
    for previous_row in range(row):
        previous_col = board[previous_row]

        if previous_col == col:
            return False

        if abs(previous_col - col) == abs(previous_row - row):
            return False

    return True

def solve_n_queens(board, row, n):
    if row == n:
        return True

    for col in range(n):
        if is_safe(board, row, col, n):
            board[row] = col

            if solve_n_queens(board, row + 1, n):
                return True

            board[row] = -1

    return False

def print_board(board, n):
    for row in range(n):
        line = []
        for col in range(n):
            line.append("Q" if board[row] == col else ".")
        print(" ".join(line))

n = int(input("Enter N for N-Queens: "))

if n < 1:
    print("N must be positive.")
else:
    board = [-1] * n

    if solve_n_queens(board, 0, n):
        print("One valid solution:")
        print_board(board, n)
    else:
        print("No solution exists.")
```

**Write a program to demonstrate the Hill Climbing Algorithm for finding the maximum value of F(x) = -x^2 + 4x.**

```python
def f(x):
    return -x**2 + 4*x

def hill_climbing(start, step=0.1):
    current = start

    while True:
        current_value = f(current)
        left = current - step
        right = current + step

        best = current

        if f(left) > f(best):
            best = left

        if f(right) > f(best):
            best = right

        if f(best) <= current_value:
            return current, current_value

        current = best

start = float(input("Enter starting value of x: "))
x, value = hill_climbing(start)

print("Local maximum x =", round(x, 4))
print("Maximum value =", round(value, 4))
```

### Q2 — Question 38

**Write a program to solve the Map Colouring Problem using the Constraint Satisfaction Problem (CSP) approach.**

```python
def is_safe(node, color, assignment, graph):
    for neighbor in graph[node]:
        if assignment.get(neighbor) == color:
            return False
    return True

def solve_csp(nodes, colors, graph, assignment):
    if len(assignment) == len(nodes):
        return True

    node = next(n for n in nodes if n not in assignment)

    for color in colors:
        if is_safe(node, color, assignment, graph):
            assignment[node] = color

            if solve_csp(nodes, colors, graph, assignment):
                return True

            del assignment[node]

    return False

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D"],
    "D": ["B", "C"]
}

colors = ["Red", "Green", "Blue"]
assignment = {}

if solve_csp(list(graph.keys()), colors, graph, assignment):
    print("Map Coloring Solution:")
    for region, color in assignment.items():
        print(region, "->", color)
else:
    print("No solution exists.")
```

**Write a program to implement the AO* Algorithm.**

```python
graph = {
    "A": [
        [("B", 1), ("C", 1)],
        [("D", 4)]
    ],
    "B": [
        [("E", 2)],
        [("F", 3)]
    ],
    "C": [
        [("G", 2)]
    ],
    "D": [
        [("H", 2)]
    ],
    "E": [],
    "F": [],
    "G": [],
    "H": []
}

heuristic = {
    "A": 6, "B": 2, "C": 2, "D": 4,
    "E": 0, "F": 0, "G": 0, "H": 0
}

solution = []

def ao_star(node):
    if not graph[node]:
        return heuristic[node]

    best_cost = float("inf")
    best_option = None

    for option in graph[node]:
        option_cost = 0

        for child, edge_cost in option:
            option_cost += edge_cost + ao_star(child)

        if option_cost < best_cost:
            best_cost = option_cost
            best_option = option

    solution.append((node, best_option, best_cost))
    heuristic[node] = best_cost

    return best_cost

start = "A"
cost = ao_star(start)

print("AO* Optimal Cost:", cost)
print("Selected AND-OR decisions:")

for node, option, option_cost in solution:
    print(node, "->", [child for child, _ in option],
          "Cost =", option_cost)
```
